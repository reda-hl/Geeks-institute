**`Convert the two following lists, into dictionaries.
Hint: Use the zip method`**

```python
keys = ['Ten', 'Twenty', 'Thirty']
values = [10, 20, 30]
```

```python
list_to_dict={}
for item in zip(keys,values):
    list_to_dict[item[0]] = item[1]

print(list_to_dict)
```
