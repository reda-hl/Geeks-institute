**`Write code that asks the user for their name and determines whether or not you have the same name, print out a funny message based on the outcome.`**

```python
Name = input('Please Enter your Name! ')
My_Name = 'Reda'
if Name.capitalize() == My_Name.capitalize():
  print("Are we clones, or did the universe just glitch? We have the same Name hahaha")
else:
  print("Your name is cool too, I guess. Mine is still easier to remember.")
```
