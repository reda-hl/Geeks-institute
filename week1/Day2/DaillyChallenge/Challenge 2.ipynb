**`Challenge 2 : Write a program that asks a string to the user, and display a new string with any duplicate consecutive letters removed.`**

```python
word = input("Choose a word string : ")

new_word = word[0]

for i in range(1,len(word)):
  if word[i] !=word[i-1]:
    new_word += word[i]

print(new_word)
```
