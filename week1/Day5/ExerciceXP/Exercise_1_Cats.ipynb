```python
class Cat:
    def __init__(self, cat_name, cat_age):
        self.name = cat_name
        self.age = cat_age
```

```python
cat1=Cat('reda',28)
cat2=Cat('hamza',32)
cat3=Cat('ahmed',15)
```

```python
def oldest(cat1,cat2,cat3):
  if cat1.age > cat2.age:
    if cat1.age > cat3.age:
      return cat1.name, cat1.age
    elif cat1.age < cat3.age:
      return cat3.name ,cat3.age
    else:
      return "They Have same Age , name : {}  age :{} and , name : {}  age :{}".format(cat1.name ,cat1.age ,cat2.name , cat2.age )
  elif cat1.age < cat2.age :
    if cat2.age > cat3.age :
      return cat2.name , cat2.age
    elif cat2.age < cat3.age :
      return cat3.name ,cat3.age
    else:
      return "They Have same Age , name : {}  age :{} and , name : {}  age :{}".format(cat3.name ,cat3.age ,cat2.name , cat2.age )

  else:
    return "They Have same Age"
```

```python
print("The oldest cat is {}, and is {} years old.".format(oldest(cat1,cat2,cat3)[0],oldest(cat1,cat2,cat3)[1]))
```
