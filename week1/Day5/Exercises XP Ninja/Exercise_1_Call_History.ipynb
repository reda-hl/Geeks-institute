Instructions





Create a class called Phone. This class takes a parameter called phone_number. When instantiating an object create an attribute called call_history which value is an empty list.




Add a method called call that takes both self and other_phone (i.e another Phone object) as parameters. The method should print a string stating who called who, and add this string to the phone’s call_history.




Add a method called show_call_history. This method should print the call_history.




Add another attribute called messages to your __init__() method which value is an empty list.




Create a method called send_message which is similar to the call method. Each message should be saved as a dictionary with the following keys:





to : the number of another Phone object



from : your phone number (also a Phone object)



content




Create the following methods: show_outgoing_messages(self), show_incoming_messages(self), show_messages_from(self)




Test your code !!!

```python
class Phone:
  def __init__(self,phone_number):
    self.phone_number = phone_number
    self.call_history=[]
    self.message = []
  def call(self,other_phone):
    mes = f"{str(other_phone.phone_number)} is calling : { str(self.phone_number)}"
    self.call_history.append(mes)
    print(mes)

  def show_call_history(self):
    for element in self.call_history:
      print(element)

  def send_message(self, other_phone , message_content):
    mess = {"to" : other_phone.phone_number,
            "from" : self.phone_number,
            "content":message_content}

    self.message.append(mess)
    print("Done !")

  def show_outgoing_messages(self):
    for element in self.message:
      print(element)

  def show_incoming_messages(self):
    for element in self.message:
      if element['to'] == self.phone_number :
        print(element)
  def show_messages_from(self , phone_number):
    for element in self.message:
      if element['from'] == phone_number :
        print(element)
```

```python
phone1 = Phone("0611111111")
phone2 = Phone("0622222222")

phone1.call(phone2)
phone2.call(phone1)

phone1.show_call_history()

phone1.send_message(phone2, "Hello!")
phone2.send_message(phone1, "Hi there!")

print("\nPhone 1 messages:")
phone1.show_outgoing_messages()

print("\nPhone 1 incoming messages:")
phone1.show_incoming_messages()

print("\nMessages from phone 1:")
phone1.show_messages_from("0611111111")
```
