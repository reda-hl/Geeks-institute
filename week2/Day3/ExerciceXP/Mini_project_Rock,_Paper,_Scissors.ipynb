```python
import random
class Game:
  def get_user_item(self):

    type= input("choisir une possibilité parmi ces trois : rock, paper ou scissors : ").lower()
    while type != "rock" and type !=  "paper" and type != "scissors":
      type= input("choisir une possibilité parmi ces trois : rock, paper ou scissors : ").lower()

    return type

  def get_computer_item(self):
    poss = ["rock", "paper", "scissors"]
    return random.choice(poss)


  def get_game_result(self, user_item, computer_item):
    if user_item =="scissors" and computer_item =="scissors" or user_item =="rock" and computer_item =="rock" or user_item =="paper" and computer_item =="paper" :
      return 'draw'

    if user_item =="rock" and computer_item =="scissors" or  user_item =="scissors" and computer_item =="paper" or  user_item =="paper" and computer_item =="rock" :
      return "win"
    else:
      return "loss"

  def play(self):
    user_item = self.get_user_item()
    computer_item = self.get_computer_item()
    result = self.get_game_result(user_item, computer_item)

    print(f"You selected : {user_item} ")
    print(f"The computer selected : {computer_item} ")
    print(f"you {result}")

    return result

def get_user_menu_choice():
  game = Game()
  print("-p : jouer  \n-s : afficher les résultats \n-q : quitter")
  choix = input("Choisir un choix : ")
  if choix == 'p':
    result =  game.play()
    return result
  elif choix == 's':
    return "s"
  elif choix =='q':
    return "q"
  else :
    print("Choix invalide")
    return None

def print_results(results):
  somme = results['win'] + results['loss'] + results['draw']
  print(f"{results['win']} victoires (win) \n{results['loss']} défaite (loss)  \n{results['draw']} matchs nuls (draw) \nTotal Game : {somme}")

def main():
  results={'win':0,'loss':0,'draw':0}
  while True:
    menu = get_user_menu_choice()
    if menu in results:
      results[menu]+=1
    elif menu =='s':
      print_results(results)

    elif menu =='q':
      print_results(results)
      break
```

```python
main()
```
